## Bivariate Analysis

Bivariate analysis examines the relationship between two variables.

### 1. Study Hours vs GPA

This analysis investigates whether there is an observable relationship between the number of hours students study per day and their GPA.

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Study_Hours_Per_Day",
    y="GPA"
)

plt.title("Study Hours vs GPA")
plt.xlabel("Study Hours Per Day")
plt.ylabel("GPA")

plt.show()
```

### Observation

The scatter plot shows an observable positive relationship between study hours per day and GPA.

Students with higher daily study hours generally tend to have higher GPA values. However, there is noticeable variation in GPA at similar study-hour levels.

This indicates an association between study time and academic performance in this dataset. The relationship should not be interpreted as proof that increased study hours directly cause higher GPA.

### 2. Sleep Hours vs GPA

This analysis examines the relationship between students' daily sleep duration and their GPA.

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Sleep_Hours_Per_Day",
    y="GPA"
)

plt.title("Sleep Hours vs GPA")
plt.xlabel("Sleep Hours Per Day")
plt.ylabel("GPA")

plt.show()
```

### Observation

The scatter plot shows that GPA values are widely distributed across different sleep durations.

Unlike the Study Hours vs GPA relationship, there is no strong and consistent upward or downward pattern visible between sleep hours and GPA. Students with similar amounts of sleep can have considerably different GPA values.

Therefore, the relationship between sleep duration and GPA appears relatively weak or unclear from this visualization.

### 3. Physical Activity Hours vs GPA

This analysis examines whether there is an observable relationship between daily physical activity and students' GPA.

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Physical_Activity_Hours_Per_Day",
    y="GPA"
)

plt.title("Physical Activity Hours vs GPA")
plt.xlabel("Physical Activity Hours Per Day")
plt.ylabel("GPA")

plt.show()
```

The scatter plot shows a slight negative relationship between physical activity hours and GPA.

Students with lower to moderate physical activity hours show a wide range of GPA values, while higher physical activity hours are somewhat more concentrated around lower GPA values.

However, the points are widely scattered, indicating that physical activity alone does not strongly explain differences in GPA. This represents an observed association in the dataset and should not be interpreted as physical activity causing lower academic performance.

## Multivariate Analysis

Multivariate analysis examines relationships among multiple numerical variables simultaneously. A correlation heatmap is used to identify the strength and direction of relationships between lifestyle factors and GPA.

```python
plt.figure(figsize=(10, 7))

correlation = df.select_dtypes(include=np.number).corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Student Lifestyle and GPA")

plt.show()
```

The correlation heatmap shows the relationships between the numerical lifestyle variables and GPA.

The strongest positive correlation with GPA is observed for Study Hours Per Day, with a correlation coefficient of approximately 0.73. This indicates a strong positive association between study time and GPA in this dataset.

Physical Activity Hours Per Day has a moderate negative correlation with GPA of approximately -0.34.

Social Hours Per Day shows a very weak negative correlation with GPA, while Extracurricular Hours Per Day and Sleep Hours Per Day have correlations close to zero.

Student_ID has almost no correlation with GPA, which is expected because it is only an identifier and does not represent a student lifestyle factor.

Correlation indicates association between variables and does not establish a causal relationship.

## Outlier Analysis

Outlier analysis is used to identify observations that are unusually high or low compared with the rest of the data.

A boxplot is used to visually identify potential outliers in GPA.

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    y="GPA"
)

plt.title("Boxplot of GPA")
plt.ylabel("GPA")

plt.show()
```

### Observation

The GPA boxplot shows that most student GPA values are concentrated around the middle of the distribution, with a median of approximately 3.1.

A few observations appear outside the lower and upper whiskers. These potential outliers occur near GPA values of approximately 2.24–2.25 and 4.00.

These observations are not necessarily errors because they fall within the valid GPA range of the dataset. They are identified as potential outliers because they are relatively far from the central distribution.

```python
Q1 = df["GPA"].quantile(0.25)
Q3 = df["GPA"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["GPA"] < lower_bound) |
    (df["GPA"] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of GPA Outliers:", len(outliers))
```

### IQR-Based Outlier Detection

The Interquartile Range (IQR) method was used to identify potential GPA outliers.

The first quartile (Q1) is 2.90 and the third quartile (Q3) is 3.33, giving an IQR of 0.43.

Using the 1.5 × IQR rule, the lower bound is 2.255 and the upper bound is 3.975.

A total of 4 GPA observations fall outside these bounds and are therefore identified as potential outliers.

These observations were not removed because an outlier is not necessarily an incorrect value. The values may represent students with unusually low or high GPA within the dataset.

```python
print("GPA Outlier Values:")
print(outliers["GPA"].values)
```

### Outlier Analysis Conclusion

Using the IQR method, four GPA observations were identified as potential outliers.

The identified GPA values are 4.00, 2.25, 2.24, and 2.25. These values lie outside the calculated IQR boundaries of 2.255 and 3.975.

Although these observations are statistically identified as potential outliers, they were not removed because they fall within the valid GPA range of the dataset. They may represent students with unusually high or low academic performance rather than data-entry errors.

Therefore, the original observations are retained for further analysis.
